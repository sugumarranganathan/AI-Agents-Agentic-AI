Create the Gemini Client

```python
from google.colab import userdata
from google import genai

# Load Secret
GEMINI_API_KEY = userdata.get("GEMINI_API_KEY_2")

# Create Gemini Client
client = genai.Client(api_key=GEMINI_API_KEY)

print("Gemini Client Ready ✅")
```

Writer Agent

```python
def writer_agent(task, feedback=""):

    prompt = f"""
You are Writer Agent.

Your job is to write Python code.

If Reviewer gives feedback,
correct the code.

Return ONLY executable Python code.

Task:
{task}

Reviewer Feedback:
{feedback}
"""

    response = client.models.generate_content(
        model="gemini-3.6-flash",
        contents=prompt
    )

    return response.text
```

Reviewer Agent

```python
def reviewer_agent(code):

    prompt = f"""
You are Reviewer Agent.

Review this Python code.

If code is correct,
reply ONLY

APPROVED

Otherwise explain what should
be fixed.

Python Code:

{code}
"""

    response = client.models.generate_content(
        model="gemini-3.6-flash",
        contents=prompt
    )

    return response.text
```

Verify the Client

```python
response = client.models.generate_content(
    model="gemini-3.6-flash",
    contents="Hello"
)

print(response.text)
```

User Task

```python
task = """
Student Marks

Tamil = 90
English = 85
Maths = 95
Science = 92
Social = 88

Calculate

Total

Average

Percentage

Print the result.
"""
```

Start Conversation

```python
feedback = ""

for round_no in range(1,4):

    print("="*60)
    print(f"ROUND {round_no}")
    print("="*60)

    print("\n🤖 Writer Agent\n")

    code = writer_agent(task, feedback)

    print(code)

    print("\n🤖 Reviewer Agent\n")

    feedback = reviewer_agent(code)

    print(feedback)

    if "APPROVED" in feedback.upper():
        print("\nConversation Finished")
        break
```
