Import Libraries

````python
from google.colab import userdata
from google import genai
````

Load Gemini Secret

````python
GEMINI_API_KEY = userdata.get("GEMINI_API_KEY_2")

print("Loaded:", GEMINI_API_KEY is not None)
````

Create Gemini Client

````python
client = genai.Client(
    api_key=GEMINI_API_KEY
)

print("Gemini Client Ready")
````

Create Writer Agent

````python
def writer_agent(task):

    system_prompt = """
You are an Expert Python Developer.

Your job is:

1. Write clean Python code.

2. Never explain.

3. Return only executable Python code.

4. Use best coding practices.
"""

    response = client.models.generate_content(
        model="gemini-3.6-flash",
        contents=f"{system_prompt}\n\nTask:\n{task}"
    )

    return response.text
````

Create Reviewer Agent

````python
def reviewer_agent(code):

    system_prompt = """
You are a Senior Python Code Reviewer.

Your job is:

1. Review the Python code.

2. Fix bugs.

3. Improve readability.

4. Return ONLY corrected executable Python code.
"""

    response = client.models.generate_content(
        model="gemini-3.6-flash",
        contents=f"{system_prompt}\n\nPython Code:\n{code}"
    )

    return response.text
````

User Task

````python
task = """
Student Marks

Tamil = 90
English = 85
Maths = 95
Science = 92
Social = 88

Calculate

1.Total

2.Average

3.Percentage

Print the results.
"""
````

Writer Agent Starts

````python
print("="*60)
print("WRITER AGENT")
print("="*60)

python_code = writer_agent(task)

print(python_code)
````

Reviewer Agent Replies

````python
print("="*60)
print("REVIEWER AGENT")
print("="*60)

reviewed_code = reviewer_agent(python_code)

print(reviewed_code)
````

Execute Final Code

````python
reviewed_code = reviewed_code.replace("```python", "")
reviewed_code = reviewed_code.replace("```", "")
reviewed_code = reviewed_code.strip()

print(reviewed_code)
````

````python
exec(reviewed_code)
````
