Question 3: Code Generator ➝ Code Reviewer ➝ Bug  Fixer
  Task: Have Agent A generate Python code, Agent B review it, and Agent C suggest bug fixes.

=====

Agent A: Generates Python code.

Agent B: Reviews the generated code for quality, style, and potential issues.

Agent C: Fixes bugs and returns the improved code.

```python
from google.colab import userdata
from groq import Groq

GROQ_API_KEY = userdata.get("GROQ_API_KEY")

client = Groq(api_key=GROQ_API_KEY)

MODEL = "llama-3.3-70b-versatile"
```

```python
def run_agent(system_prompt, user_message):

    response = client.chat.completions.create(
        model=MODEL,
        temperature=0.2,
        messages=[
            {
                "role":"system",
                "content":system_prompt
            },
            {
                "role":"user",
                "content":user_message
            }
        ]
    )

    return response.choices[0].message.content
```

Agent A – Code Generator

```python
generator_prompt = """
You are a Python Code Generator.

Task:
- Generate Python code only.
- Do not explain.
- Produce clean and readable code.
"""

task = """
Write a Python function that checks whether a number is prime.
"""

generated_code = run_agent(generator_prompt, task)

print("========== GENERATED CODE ==========\n")
print(generated_code)
```

Agent B – Code Reviewer

```python
review_prompt = """
You are a Senior Python Code Reviewer.

Review the given code.

Check for:
- Bugs
- Code quality
- Naming
- Readability
- Efficiency

Return review comments only.
"""

review = run_agent(
    review_prompt,
    generated_code
)

print("\n========== REVIEW ==========\n")
print(review)
```

Agent C – Bug Fixer

```python
bugfix_prompt = """
You are a Python Bug Fixer.

You will receive:

1. Python code
2. Review comments

Return the corrected and improved Python code only.
"""

bugfix_input = f"""
Generated Code:

{generated_code}

Reviewer Comments:

{review}
"""

fixed_code = run_agent(
    bugfix_prompt,
    bugfix_input
)

print("\n========== FIXED CODE ==========\n")
print(fixed_code)
```
