Question 2: Language Translator ➝ Culture Rephraser

Task: Use one agent to translate from English to French, then another to culturally rephrase the French text.


Agent 1 (Translator): Translates text from English to French.

Agent 2 (Culture Rephraser): Rewrites the French text to sound natural, fluent, and culturally appropriate for native French speakers.

    English Text
          │
          ▼
    Translator Agent
    (English → French)
          │
          ▼
    French Translation
          │
          ▼
    Culture Rephraser Agent
    (Native French Style)
          │
          ▼
    Final Culturally Adapted French Text

```python
from google.colab import userdata
from groq import Groq

GROQ_API_KEY = userdata.get("GROQ_API_KEY")

client = Groq(api_key=GROQ_API_KEY)

MODEL = "llama-3.3-70b-versatile"
```

Generic Agent Function

```python
def run_agent(system_prompt, user_message):
    response = client.chat.completions.create(
        model=MODEL,
        temperature=0.2,
        messages=[
            {
                "role": "system",
                "content": system_prompt
            },
            {
                "role": "user",
                "content": user_message
            }
        ]
    )

    return response.choices[0].message.content
```

Translator Agent

```python
translator_prompt = """
You are a professional translator.

Task:
- Translate English into French.
- Keep the meaning accurate.
- Return ONLY the French translation.
"""

english_text = """
Welcome to our restaurant.
We serve fresh food every day.
Thank you for visiting us.
"""

french_translation = run_agent(
    translator_prompt,
    english_text
)

print("========== TRANSLATOR OUTPUT ==========\n")
print(french_translation)
```

Culture Rephraser Agent

```python
culture_prompt = """
You are a French cultural language expert.

Task:
- Read the French text.
- Rewrite it so it sounds natural to native French speakers.
- Preserve the original meaning.
- Return ONLY the improved French text.
"""

culturally_rephrased = run_agent(
    culture_prompt,
    french_translation
)

print("\n========== CULTURAL REPHRASER OUTPUT ==========\n")
print(culturally_rephrased)
```

Translator Agent converts English → French.

Culture Rephraser Agent makes the French sound more natural and culturally appropriate for native French speakers.
