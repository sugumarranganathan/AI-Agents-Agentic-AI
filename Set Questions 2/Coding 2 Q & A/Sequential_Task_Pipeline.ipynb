Question: Sequential Task Pipeline

Task: Implement a 4-step document workflow using Gemini-powered ConversableAgents. The steps are:

DraftWriter → Writes a short article draft on a given topic.

EditorAgent → Refines the grammar, structure, and tone of the draft.

PlagiarismChecker → Checks if the content is original or copied.

PublisherAgent → Finalizes the article and prepares it for publishing.

Each agent should take the previous agent's output and continue the task without losing context.  Print each step’s output clearly.
Use generate_reply() for all interactions.

Create 4 Conversable Agents

```python
DraftWriter = ConversableAgent(
    name="DraftWriter",
    system_message="""
You are a professional article writer.

Your job:
- Write a short article.
- Keep it around 180 words.
- Use a clear introduction, body, and conclusion.
""",
    llm_config=llm_config,
    human_input_mode="NEVER",
)

EditorAgent = ConversableAgent(
    name="EditorAgent",
    system_message="""
You are an experienced editor.

Improve:
- Grammar
- Sentence structure
- Readability
- Professional tone

Do not change the article meaning.
""",
    llm_config=llm_config,
    human_input_mode="NEVER",
)

PlagiarismChecker = ConversableAgent(
    name="PlagiarismChecker",
    system_message="""
You are a plagiarism reviewer.

Analyze the article.

State:
1. Whether it appears original.
2. Mention if there are any commonly used phrases.
3. Give an originality score out of 100.

Finally provide the same article unchanged.
""",
    llm_config=llm_config,
    human_input_mode="NEVER",
)

PublisherAgent = ConversableAgent(
    name="PublisherAgent",
    system_message="""
You are a publishing manager.

Finalize the article.

Provide:

Title

Final Article

Publishing Status:
Ready for Publishing
""",
    llm_config=llm_config,
    human_input_mode="NEVER",
)
```

Sequential Task Pipeline using generate_reply()

```python
topic = "Benefits of Artificial Intelligence in Healthcare"

# ============================================================
# STEP 1 : Draft Writer
# ============================================================

draft = DraftWriter.generate_reply(
    messages=[
        {
            "role": "user",
            "content": f"Write a short article about: {topic}"
        }
    ]
)

print("="*70)
print("STEP 1 : DRAFT WRITER")
print("="*70)
print(draft)



# ============================================================
# STEP 2 : Editor
# ============================================================

edited = EditorAgent.generate_reply(
    messages=[
        {
            "role": "user",
            "content": draft
        }
    ]
)

print("\n")
print("="*70)
print("STEP 2 : EDITOR AGENT")
print("="*70)
print(edited)



# ============================================================
# STEP 3 : Plagiarism Checker
# ============================================================

checked = PlagiarismChecker.generate_reply(
    messages=[
        {
            "role": "user",
            "content": edited
        }
    ]
)

print("\n")
print("="*70)
print("STEP 3 : PLAGIARISM CHECKER")
print("="*70)
print(checked)



# ============================================================
# STEP 4 : Publisher
# ============================================================

published = PublisherAgent.generate_reply(
    messages=[
        {
            "role": "user",
            "content": checked
        }
    ]
)

print("\n")
print("="*70)
print("STEP 4 : PUBLISHER")
print("="*70)
print(published)
```
