Question 3. Role-Based Chat (Specialized Roles)

Question: Assign roles to agents:

DoctorAgent gives medical insight
DietitianAgent suggests a meal plan
FitnessCoachAgent suggests workouts For a given user query like: “Help me manage diabetes naturally.”


```python
!pip -q install pyautogen groq
```

```python
import os
from google.colab import userdata

os.environ["GROQ_API_KEY"] = userdata.get("GROQ_API_KEY")
```

```python
llm_config = {
    "config_list": [
        {
            "model": "llama-3.3-70b-versatile",
            "base_url": "https://api.groq.com/openai/v1",
            "api_key": os.environ["GROQ_API_KEY"],
        }
    ],
    "temperature": 0.3,
}
```

Create Specialized Agents

```python
!pip install pyautogen==0.2.35
```

```python
import os
from google.colab import userdata
from autogen import ConversableAgent
```

Create Specialized Agents

```python
doctor = ConversableAgent(
    name="DoctorAgent",
    system_message="""
You are a licensed medical doctor.

Responsibilities:
- Explain diabetes in simple language.
- Suggest natural ways to manage diabetes.
- Mention lifestyle modifications.
- Advise consulting a physician before changing medication.

Do not prescribe medicines.
""",
    llm_config=llm_config,
    human_input_mode="NEVER",
)

dietitian = ConversableAgent(
    name="DietitianAgent",
    system_message="""
You are a certified dietitian.

Responsibilities:
- Create a healthy Indian meal plan.
- Recommend diabetic-friendly foods.
- Suggest foods to avoid.
- Include breakfast, lunch, snacks and dinner.
""",
    llm_config=llm_config,
    human_input_mode="NEVER",
)

fitness = ConversableAgent(
    name="FitnessCoachAgent",
    system_message="""
You are a certified fitness coach.

Responsibilities:
- Recommend beginner-friendly workouts.
- Suggest walking, yoga and stretching.
- Mention workout duration.
- Include safety precautions.
""",
    llm_config=llm_config,
    human_input_mode="NEVER",
)
```

User Query

```python
query = "Help me manage diabetes naturally."
```

Doctor Agent

```python
doctor_reply = doctor.generate_reply(
    messages=[
        {
            "role": "user",
            "content": query
        }
    ]
)

print("="*70)
print("DOCTOR AGENT")
print("="*70)
print(doctor_reply)
```

Dietitian Agent

```python
diet_reply = dietitian.generate_reply(
    messages=[
        {
            "role": "user",
            "content": query
        }
    ]
)

print("="*70)
print("DIETITIAN AGENT")
print("="*70)
print(diet_reply)
```

Fitness Coach Agent

```python
fitness_reply = fitness.generate_reply(
    messages=[
        {
            "role": "user",
            "content": query
        }
    ]
)

print("="*70)
print("FITNESS COACH AGENT")
print("="*70)
print(fitness_reply)
```

Final Combined Health Plan

```python
print("\n")
print("="*80)
print("PERSONALIZED DIABETES MANAGEMENT PLAN")
print("="*80)

print("\nUser Query")
print(query)

print("\n")
print("="*80)
print("DOCTOR'S ADVICE")
print("="*80)
print(doctor_reply)

print("\n")
print("="*80)
print("DIET PLAN")
print("="*80)
print(diet_reply)

print("\n")
print("="*80)
print("FITNESS PLAN")
print("="*80)
print(fitness_reply)
```
