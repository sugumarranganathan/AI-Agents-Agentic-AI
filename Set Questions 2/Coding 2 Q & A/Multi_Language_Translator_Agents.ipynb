```python
import os
from google.colab import userdata
from autogen import ConversableAgent
```

Create Translator Agents


English → Spanish Agent

```python
import os
from google.colab import userdata

os.environ["GROQ_API_KEY"] = userdata.get("GROQ_API_KEY")

llm_config = {
    "config_list": [
        {
            "model": "llama-3.3-70b-versatile",
            "base_url": "https://api.groq.com/openai/v1",
            "api_key": os.environ["GROQ_API_KEY"],
        }
    ],
    "temperature": 0.2,
}
```

Create Translator Agents

English → Spanish Agent

```python
english_to_spanish = ConversableAgent(
    name="EnglishToSpanish",
    system_message="""
You are a professional translator.

Translate the given English paragraph into Spanish.

Return only the Spanish translation.
""",
    llm_config=llm_config,
    human_input_mode="NEVER",
)
```

Spanish → French Agent

```python
spanish_to_french = ConversableAgent(
    name="SpanishToFrench",
    system_message="""
You are a professional translator.

Translate the given Spanish paragraph into French.

Return only the French translation.
""",
    llm_config=llm_config,
    human_input_mode="NEVER",
)
```

French → German Agent

```python
french_to_german = ConversableAgent(
    name="FrenchToGerman",
    system_message="""
You are a professional translator.

Translate the given French paragraph into German.

Return only the German translation.
""",
    llm_config=llm_config,
    human_input_mode="NEVER",
)
```

Input Paragraph

```python
paragraph = """
Artificial Intelligence is transforming healthcare by improving diagnosis,
enhancing patient care, and enabling doctors to make better decisions.
"""
```

English → Spanish

```python
spanish_reply = english_to_spanish.generate_reply(
    messages=[
        {
            "role": "user",
            "content": paragraph
        }
    ]
)

print("="*70)
print("ENGLISH TO SPANISH")
print("="*70)
print(spanish_reply)
```

Spanish → French

```python
french_reply = spanish_to_french.generate_reply(
    messages=[
        {
            "role": "user",
            "content": spanish_reply
        }
    ]
)

print("="*70)
print("SPANISH TO FRENCH")
print("="*70)
print(french_reply)
```

French → German

```python
german_reply = french_to_german.generate_reply(
    messages=[
        {
            "role": "user",
            "content": french_reply
        }
    ]
)

print("="*70)
print("FRENCH TO GERMAN")
print("="*70)
print(german_reply)
```

Final Output

```python
print("\n")
print("="*80)
print("FINAL TRANSLATION")
print("="*80)

print("\nOriginal English Paragraph:")
print(paragraph)

print("\nSpanish Translation:")
print(spanish_reply)

print("\nFrench Translation:")
print(french_reply)

print("\nFinal German Translation:")
print(german_reply)
```

| **Code**                                 | **Purpose**    | **Description**                                                |
| ---------------------------------------- | -------------- | -------------------------------------------------------------- |
| `print("\n")`                            | New Line       | Prints a blank line to improve readability.                    |
| `print("="*80)`                          | Separator      | Prints 80 equal (`=`) symbols as a horizontal line.            |
| `print("FINAL TRANSLATION")`             | Heading        | Displays the title of the final output section.                |
| `print("="*80)`                          | Separator      | Prints another horizontal line below the heading.              |
| `print("\nOriginal English Paragraph:")` | Section Title  | Displays the heading for the original English input.           |
| `print(paragraph)`                       | Display Input  | Prints the original English paragraph entered by the user.     |
| `print("\nSpanish Translation:")`        | Section Title  | Displays the heading for the Spanish translation.              |
| `print(spanish_reply)`                   | Display Output | Prints the output generated by the **EnglishToSpanish** agent. |
| `print("\nFrench Translation:")`         | Section Title  | Displays the heading for the French translation.               |
| `print(french_reply)`                    | Display Output | Prints the output generated by the **SpanishToFrench** agent.  |
| `print("\nFinal German Translation:")`   | Section Title  | Displays the heading for the final German translation.         |
| `print(german_reply)`                    | Display Output | Prints the output generated by the **FrenchToGerman** agent.   |
