```python
from autogen import ConversableAgent

planner = ConversableAgent(
    name="PlannerAgent",

    system_message="""
You are a Task Planning Agent.

Responsibilities:
1. Understand the user's goal.
2. Break the task into logical steps.
3. Return only the execution plan.
""",

    llm_config=config,

    human_input_mode="NEVER",
)
```

Worker Agent

```python
worker = ConversableAgent(
    name="WorkerAgent",

    system_message="""
You are a Task Execution Agent.

Responsibilities:
1. Read the task plan.
2. Explain how to execute each step.
3. Return a practical execution guide.
""",

    llm_config=config,

    human_input_mode="NEVER",
)
```

User Task

```python
task = "Plan steps to clean and organize my digital files."
```

Planner Creates Plan

```python
task_plan = planner.generate_reply(
    messages=[
        {
            "role": "user",
            "content": task
        }
    ]
)

print("="*60)
print("TASK PLAN")
print("="*60)
print(task_plan)
```

Worker Executes Plan

```python
execution = worker.generate_reply(
    messages=[
        {
            "role": "user",
            "content": task_plan
        }
    ]
)

print("\n"+"="*60)
print("EXECUTION GUIDE")
print("="*60)
print(execution)
```
