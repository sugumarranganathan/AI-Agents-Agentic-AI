Create Meeting Summary Agent

```python
from autogen import ConversableAgent

summarizer = ConversableAgent(
    name="MeetingBot",

    system_message="""
You are a Meeting Summary Agent.

Your responsibilities are:

1. Read the meeting notes carefully.
2. Summarize the discussion.
3. Extract the important decisions.
4. Extract action items.
5. Mention deadlines if available.
6. Return the summary in professional bullet points.
""",

    llm_config=config,

    human_input_mode="NEVER",
)
```

Meeting Notes

```python
meeting_notes = """
We discussed quarterly targets,
team expansion,
marketing campaigns.

The deadline for reports is April 30th.

The HR team will recruit five new software engineers.

The marketing team will launch social media campaigns next month.

Weekly review meetings will be conducted every Monday.
"""
```

Generate Summary

```python
summary = summarizer.generate_reply(
    messages=[
        {
            "role": "user",
            "content": meeting_notes
        }
    ]
)

print(summary)
```
