Create Interview Agent

```python
from autogen import ConversableAgent

interview_bot = ConversableAgent(
    name="InterviewBot",

    system_message="""
You are an AI Interview Assistant.

Your responsibilities are:

1. Ask professional interview questions.
2. Evaluate the candidate's answer.
3. Give constructive feedback.
4. Suggest improvements.
5. Keep the interview professional.
6. Return only the interview response.
""",

    llm_config=config,

    human_input_mode="NEVER",
)
```

Interview Question

```python
question = "Tell me about your strengths and weaknesses."
```

Generate Interview Response

```python
result = interview_bot.generate_reply(
    messages=[
        {
            "role": "user",
            "content": question
        }
    ]
)

print(result)
```

```python
from autogen import ConversableAgent

interview_bot = ConversableAgent(
    name="InterviewBot",

    system_message="""
You are an AI HR Interview Assistant.

Your task is to answer interview questions as an experienced candidate
and then evaluate that answer.

Return your response ONLY in the following format.

Question:
<repeat the interview question>

Candidate Answer:
<professional interview answer>

Evaluation Score:
<Score out of 10>

Strengths:
• Point 1
• Point 2
• Point 3

Areas for Improvement:
• Point 1
• Point 2
• Point 3

HR Feedback:
<Short HR feedback>

Final Recommendation:
<Select ONE>

Recommended
Recommended with Improvement
Not Recommended

Do not add any extra explanation.
""",

    llm_config=config,

    human_input_mode="NEVER",
)
```

```python
question = "Tell me about your strengths and weaknesses."
```

```python
response = interview_bot.generate_reply(
    messages=[
        {
            "role": "user",
            "content": question
        }
    ]
)

print(response)
```
