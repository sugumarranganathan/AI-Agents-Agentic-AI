```python
config = {
    "config_list": [
        {
            "model": "llama-3.3-70b-versatile",
            "api_key": GROQ_API_KEY,
            "base_url": "https://api.groq.com/openai/v1",
            "price": [0.59, 0.79]
        }
    ],
    "temperature": 0.2,
    "timeout": 120,
}
```

Create Translator Agent

```python
from autogen import ConversableAgent

translator = ConversableAgent(
    name="Translator",

    system_message="""
You are a Professional Language Translator.

Responsibilities:

1. Detect the input language.
2. Translate it into English.
3. Preserve the original meaning.
4. Use natural and fluent English.
5. Return ONLY the translated text.
""",

    llm_config=config,

    human_input_mode="NEVER",
)
```

Input Text

```python
text = "Bonjour, je m'appelle Marie."
```

Generate Translation

```python
response = translator.generate_reply(
    messages=[
        {
            "role": "user",
            "content": text
        }
    ]
)

print(response)
```
