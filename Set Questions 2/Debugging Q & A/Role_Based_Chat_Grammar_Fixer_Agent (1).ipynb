Grammar Fixer Agent

```python
from autogen import ConversableAgent

grammar_agent = ConversableAgent(
    name="GrammarFixer",

    system_message="""
You are a Grammar Fixer Agent.

Your responsibilities:

1. Correct grammar.
2. Correct spelling mistakes.
3. Improve sentence structure.
4. Keep the original meaning.
5. Return ONLY the corrected sentence.
""",

    llm_config=config,

    human_input_mode="NEVER",
)
```

Input Sentence

```python
sentence = "he dont has no idea about the matter."
```

Generate Reply

```python
response = grammar_agent.generate_reply(
    messages=[
        {
            "role": "user",
            "content": sentence
        }
    ]
)

print(response)
```
