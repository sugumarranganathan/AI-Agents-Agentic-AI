Create LLM Configuration for Groq

```python
llm_config = {
    "config_list": [
        {
            "model": "llama-3.3-70b-versatile",
            "api_key": GROQ_API_KEY,
            "base_url": "https://api.groq.com/openai/v1"
        }
    ],
    "temperature": 0.3,
    "timeout": 120,
}
```

Create the ConversableAgent

```python
from autogen import ConversableAgent

toneAgent = ConversableAgent(
    name="ToneAgent",
    llm_config=llm_config,
    human_input_mode="NEVER",
)
```

Create the Tone agent

```python
from autogen import ConversableAgent

toneAgent = ConversableAgent(
    name="ToneAgent",

    system_message="""
You are a Professional Tone Improvement Agent.

Your job is to rewrite messages into professional business English.

Rules:
- Improve grammar.
- Improve politeness.
- Keep the same meaning.
- Return only the rewritten message.
""",

    llm_config=llm_config,

    human_input_mode="NEVER",
)
```

```python
Test the Agent
```

```python
message = "can u send me the file soon?"

reply = toneAgent.generate_reply(
    messages=[
        {
            "role": "user",
            "content": message
        }
    ]
)

print(reply)
```
