Create Summary Agent

```python
from autogen import ConversableAgent

summarizer = ConversableAgent(
    name="Summarizer",

    system_message="""
You are a Meeting and Review Summarizer.

Read the text carefully.

Return only a concise summary.

Do not explain anything.
""",

    llm_config=config,

    human_input_mode="NEVER",
)
```

Create Sentiment Agent

```python
sentimenter = ConversableAgent(
    name="SentimentAnalyzer",

    system_message="""
You are a Sentiment Analysis Agent.

Analyze the given text.

Return ONLY:

Sentiment:
Positive / Neutral / Negative

Confidence:
0-100%

Reason:
One sentence.
""",

    llm_config=config,

    human_input_mode="NEVER",
)
```

Review Text

```python
review_text = """
I ordered a laptop last week.

It arrived with a broken screen.

Extremely disappointed!
"""
```

Agent Chaining

```python
summary = summarizer.generate_reply(
    messages=[
        {
            "role": "user",
            "content": review_text
        }
    ]
)

print("="*60)
print("SUMMARY")
print("="*60)
print(summary)

sentiment = sentimenter.generate_reply(
    messages=[
        {
            "role": "user",
            "content": summary
        }
    ]
)

print("\n"+"="*60)
print("SENTIMENT")
print("="*60)
print(sentiment)
```
